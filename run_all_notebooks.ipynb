# Run all notebooks in the repo

```python
# Suppress jupyter notebook warnining
import warnings
warnings.filterwarnings('ignore')
```

```python
import nbformat
import time
from nbconvert.preprocessors import ExecutePreprocessor
from pathlib import Path
```

```python
def execute_notebook(notebook_path):
    """Reads, executes, and saves a Jupyter notebook."""
    print(f"Starting: {notebook_path.name}...")
    
    # Read the notebook
    with open(notebook_path, 'r', encoding='utf-8') as f:
        nb = nbformat.read(f, as_version=4)

    # Configure the executor (adjust timeout in seconds as needed)
    ep = ExecutePreprocessor(timeout=600, kernel_name='python3')

    try:
        # Execute the notebook. The 'path' metadata allows relative imports 
        # and data loading within the notebook to work correctly.
        ep.preprocess(nb, {'metadata': {'path': notebook_path.parent}})
        
        # Save the executed notebook (overwrites the original)
        with open(notebook_path, 'w', encoding='utf-8') as f:
            nbformat.write(nb, f)
            
        print(f"✓ Successfully completed: {notebook_path.name}")
        
    except Exception as e:
        print(f"✗ Error executing {notebook_path.name}:\n{e}")
```

```python
def run_all_notebooks_in_dir(parent_directory, pause_seconds=2):
    """Recursively finds and runs all notebooks in a directory with a pause."""
    parent_path = Path(parent_directory)
    
    # 1. Find all .ipynb files recursively
    all_notebook_files = parent_path.rglob('*.ipynb')
    
    # 2. Filter out Jupyter's hidden checkpoint files first
    valid_notebooks = [nb for nb in all_notebook_files if '.ipynb_checkpoints' not in str(nb)]
    
    total_notebooks = len(valid_notebooks)
    print(f"Found {total_notebooks} notebooks to execute.\n")
    print("-" * 30)
    
    # 3. Execute with a pause
    for i, nb_file in enumerate(valid_notebooks):
        execute_notebook(nb_file)
        
        # Pause between runs, but skip the pause after the final notebook
        if i < total_notebooks - 1:
            print(f"Pausing for {pause_seconds} seconds...\n")
            time.sleep(pause_seconds)
```

# Run experiment notebooks

```python
# Define your parent directory here ('.' means the current directory)
TARGET_DIR = './notebooks/experiments' 

print(f"Searching for notebooks in: {Path(TARGET_DIR).resolve()}")
run_all_notebooks_in_dir(TARGET_DIR)
print("Done!")
```
