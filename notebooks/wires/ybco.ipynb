```python
# 2021solovyovYBCOonKaptonMaterialHighDensity
# https://www.osti.gov/servlets/purl/1764583

YBCO_conductivity = 12 # W per sqm for 1 m long cable between 50K and 4K # overestimation

width = 1E-3 # line width
thickness = 100E-6 + 1E-6 #100um(kapton) + 1um(YBCO film)
cable_area = width * thickness
```

```python
from library.fridges import TEMP_STAGES, FRIDGE_LIBRARY
# Assuming XLD1000sl
# We will use the operating temperatures and cable lengths corresponding to an XLD1000sl fridge
fridge = FRIDGE_LIBRARY["XLD1000SL"]
cooling_power_budget = fridge["cooling_power"]
operating_temp = fridge["temp"]
flange_separation = fridge["lengths"]
```

```python
print("Conductivity for YBCO")
for temp_stage in TEMP_STAGES[1:3]:
    # get cable length
    cable_length = flange_separation[temp_stage] * 1E-2 # cm --> m
    PHL = YBCO_conductivity * cable_area / cable_length 
    print(f"{temp_stage}:\t {PHL:.2e} Watts")
```
